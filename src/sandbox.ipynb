```python
import numpy as np
import pandas as pd
import plotly.express as px
# import matplotlib.pyplot as plt
# %config InlineBackend.figure_format = 'retina'

import uproot
import awkward as ak

import csv
```

```python
df = pd.read_csv("branch_metadata.csv")

df_jets = df[df["file"].str.contains("DAOD")]
df_muons = df[df["file"].str.contains("ODEO")]

df_jets = df_jets.groupby("branch_category")[["compressed_bytes", "uncompressed_bytes"]].sum().reset_index()
df_jets["compression_ratio"] = df_jets["uncompressed_bytes"] / df_jets["compressed_bytes"]
df_jets["size_reduction (%)"] = (1 - (df_jets["compressed_bytes"] / df_jets["uncompressed_bytes"])) * 100

# Bar chart of compressed vs uncompressed size for jets
fig_jets = px.bar(df_jets, x="branch_category", y=["compressed_bytes", "uncompressed_bytes"], barmode="group", color="branch_category")
fig_jets.show()

# Bar chart of compression ratio for jets
fig_jets_ratio = px.bar(df_jets, x="branch_category", y="compression_ratio", color="branch_category")
fig_jets_ratio.show()

```
